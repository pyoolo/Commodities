# Clean Spark Spread Research

Baseline notebook for the second Commodities Quant Research building block.


```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

from clean_spark.data import make_synthetic_market_data
from clean_spark.spread import clean_spark_spread, rolling_zscore, trading_position
from clean_spark.backtest import run_backtest
from clean_spark.analytics import performance_summary
```

```python
market = make_synthetic_market_data()
css = clean_spark_spread(market['power'], market['gas'], market['carbon'], 0.50, 0.36)
z = rolling_zscore(css, 30)
position = trading_position(z, 1.0)
bt = run_backtest(css, position, transaction_cost_per_unit=0.05)
performance_summary(bt)
```
